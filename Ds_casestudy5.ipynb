```python
# GPS Navigation using Weighted Graph
# Dijkstra's Shortest Path Algorithm

import heapq

# City map with distances in kilometres
graph = {
    "College": [("Hospital", 2), ("Market", 5)],
    "Hospital": [("College", 2), ("Market", 1),
                 ("Bus Stand", 4)],
    "Market": [("College", 5), ("Hospital", 1),
               ("Railway Station", 3), ("Bus Stand", 2)],
    "Bus Stand": [("Hospital", 4), ("Market", 2),
                  ("Railway Station", 2)],
    "Railway Station": [("Market", 3), ("Bus Stand", 2)]
}


def shortest_path(graph, start, destination):
    distances = {city: float("inf") for city in graph}
    previous = {city: None for city in graph}

    distances[start] = 0
    priority_queue = [(0, start)]

    while priority_queue:
        current_distance, current_city = heapq.heappop(
            priority_queue
        )

        if current_distance > distances[current_city]:
            continue

        if current_city == destination:
            break

        for neighbour, weight in graph[current_city]:
            new_distance = current_distance + weight

            if new_distance < distances[neighbour]:
                distances[neighbour] = new_distance
                previous[neighbour] = current_city

                heapq.heappush(
                    priority_queue,
                    (new_distance, neighbour)
                )

    if distances[destination] == float("inf"):
        print("No route available!")
        return

    # Reconstruct the shortest path
    path = []
    city = destination

    while city is not None:
        path.append(city)
        city = previous[city]

    path.reverse()

    print("\nStarting Location:", start)
    print("Destination:", destination)
    print("Shortest Route:", " -> ".join(path))
    print("Total Distance:", distances[destination], "km")


# Display all city locations
print("Available Locations:")
for city in graph:
    print("-", city)

# Find shortest route
start = "College"
destination = "Railway Station"

shortest_path(graph, start, destination)
```
